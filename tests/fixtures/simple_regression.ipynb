# Simple Linear Regression
Notebook fixture for parser tests.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df = pd.DataFrame({"x": [1, 2, 3], "y": [2, 4, 6]})
X = df[["x"]]
y = df["y"]
```

```python
model = LinearRegression()
model.fit(X, y)
print("score=1.0")
```
