# Junior regression experiment

This notebook intentionally keeps the pipeline in global scope, repeats preprocessing, and uses hardcoded paths and hyperparameters. The dataset is generated locally so the final metric is deterministic.

```python
import random

random.seed(2026)
DATA_PATH = "/data/hardcoded/junior_regression.csv"
MODEL_PATH = "/models/hardcoded/junior_model.bin"
LEARNING_RATE = 0.01
EPOCHS = 1_000
SEED = 2026

print(f"seed={SEED}, epochs={EPOCHS}, data={DATA_PATH}")
```

```python
# The raw data is procedural and never downloaded from the hardcoded path.
raw_x = [index / 10.0 for index in range(30)]
raw_y = [2.75 * value + 1.25 + random.uniform(-0.15, 0.15) for value in raw_x]

split_at = 21
train_raw = list(zip(raw_x[:split_at], raw_y[:split_at]))
test_raw = list(zip(raw_x[split_at:], raw_y[split_at:]))
df = train_raw + test_raw
print(f"rows={len(df)}, train={len(train_raw)}, test={len(test_raw)}")
```

```python
# Test preprocessing is intentionally duplicated instead of shared.
test_x = [row[0] / 3.0 for row in test_raw]
test_y = [row[1] for row in test_raw]
X = test_x
x2 = [value * value for value in X]
tmp = sum(x2)
print(f"test features={len(test_x)}, test scale={tmp:.3f}")
```

```python
# Training preprocessing repeats the same transformation with new globals.
train_x = [row[0] / 3.0 for row in train_raw]
train_y = [row[1] for row in train_raw]
X = train_x
x2 = [value * value for value in X]
tmp = sum(x2)
print(f"train features={len(train_x)}, train scale={tmp:.3f}")
```

```python
# Fit a real one-variable linear regression in global scope.
mean_x = sum(train_x) / len(train_x)
mean_y = sum(train_y) / len(train_y)
centered_x = [value - mean_x for value in train_x]
centered_y = [value - mean_y for value in train_y]
coefficient = sum(a * b for a, b in zip(centered_x, centered_y)) / sum(
    value * value for value in centered_x
)
intercept = mean_y - coefficient * mean_x
train_predictions = [intercept + coefficient * value for value in train_x]
train_mse = sum(
    (actual - predicted) ** 2 for actual, predicted in zip(train_y, train_predictions)
) / len(train_y)
print(f"fit coefficient={coefficient:.6f}, intercept={intercept:.6f}")
```

```python
# Final evaluation prints the deterministic target metric.
test_predictions = [intercept + coefficient * value for value in test_x]
final_mse = sum(
    (actual - predicted) ** 2 for actual, predicted in zip(test_y, test_predictions)
) / len(test_y)
print(f"model_path={MODEL_PATH}")
print(f"final_mse={final_mse:.12f}")
```
